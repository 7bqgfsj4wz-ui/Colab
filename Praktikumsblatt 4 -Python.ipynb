# Aufgabe 3

```python
def naive_Funktion(n):
    s = 0   #variable startet mit 0
    for i in range(1, n + 1):
        s += i  #Wert von i wird durch s addiert
    return s    #berechnete Summe wiedergeben

```

# Aufgabe 4

```python
import time  #Laufzeit messen
import matplotlib.pyplot as plt #Diagramm zeichnen

def summe_naiv(n):
    s = 0
    for i in range(1, n + 1):
        s += i
    return s #Summe zurückgeben

ns = [10**k for k in range(1, 7)]
zeiten = []

for n in ns:
    start = time.time() #start
    summe_naiv(n)  #eingegebene Funktion
    ende = time.time() #ende
    zeiten.append(ende - start)

plt.plot(ns, zeiten, marker="o")
plt.xlabel("n")
plt.ylabel("Laufzeit (Sekunden)")
plt.title("Laufzeit der naiven Summenfunktion")
plt.xscale("log")
plt.yscale("log")
plt.show()

```

# Aufgabe 5

```python
import time
import matplotlib.pyplot as plt

def doppelte_summe(n):
    s = 0
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            s += j
    return s

# Verschiedene Werte für n
ns = [100, 200, 400, 600, 800, 1000]
zeiten = []

for n in ns:
    start = time.time()
    doppelte_summe(n)
    ende = time.time()
    zeiten.append(ende - start)

# Darstellung
plt.plot(ns, zeiten, marker="o")
plt.xlabel("n")
plt.ylabel("Laufzeit in Sekunden")
plt.title("Laufzeit der doppelten Summe")
plt.grid(True)
plt.show()

```

# Aufgabe 6

```python
def fib_folge(n):
    a, b = 0, 1  #a aktuelle / b nächste Fib Zahl
    for _ in range(n):
        a, b = b, a + b
    return a # gibt die Fibonacci Zahl zurück

print(fib_folge(10))
print(fib_folge(100))

```

# Aufgabe 7

```python
def natuerliche_Zahl(n):
    folge = [n]

    while n != 1:
        if n % 2 == 0:      # n ist gerade
            n = n // 2
        else:               # n ist ungerade
            n = 3 * n + 1
        folge.append(n)

    return folge

startwerte = [6, 11, 19, 27]

for n in startwerte:
    f = natuerliche_Zahl(n)
    print(f"Startwert {n}:")
    print(f"  Folge: {f}")
    print(f"  Länge der Folge: {len(f)}\n")
```
